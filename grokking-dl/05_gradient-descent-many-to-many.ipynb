```python
# Код для обучения модели с множественным входом с помощью градиентного спуска

def ele_mul(number, vector):
    return [number * v for v in vector]


# Данные: количество игр, среднее количество побед, среднее количество фанатов
toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

# Истинные результаты (целевая переменная)
win_or_lose_binary = [1, 1, 0, 1]

# Инициализация весов
weights = [0.1, 0.2, 0]

# Скорость обучения
alpha = 0.01

# Количество эпох
epochs = 1000

# Обучение модели
for epoch in range(epochs):
    total_error = 0
    for i in range(len(toes)):
        # Формируем входной вектор
        input = [toes[i], wlrec[i], nfans[i]]
        true = win_or_lose_binary[i]

        # Прямой проход (предсказание)
        pred = sum(input[j] * weights[j] for j in range(len(weights)))

        # Ошибка
        error = (pred - true) ** 2
        total_error += error

        # Градиент
        delta = pred - true
        weight_deltas = ele_mul(delta, input)

        # Обновление весов
        for j in range(len(weights)):
            weights[j] -= alpha * weight_deltas[j]

    print(f"Epoch {epoch + 1}, Error: {total_error:.5f}, Weights: {weights}")
```
