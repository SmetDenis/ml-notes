```python
np.random.seed(42)

x = Tensor(np.array([1, 2, 1, 2]), autograd=True)
y = Tensor(np.array([[0], [1], [0], [1]]), autograd=True)

model = Sequential([
    Embedding(vocab_size=5, dim=3),
    Tanh(),
    Linear(3, 1),
    Sigmoid()
])
loss = MSELoss()
optimizer = SGD(params=model.get_params(), alpha=0.5)

for i in range(10):  # epochs
    y_hat = model.forward(x)
    l = loss.forward(y_hat, y)
    l.backward(Tensor(np.ones_like(l.data)))
    optimizer.step()
    print(l.data[0])
```
