```python
import torch
import torch.nn as nn
```

# Mean Squared Error (MSE)

MSE (Mean Squared Error) — это среднеквадратичная ошибка, широко используемая функция потерь для задач регрессии.

$$
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения,
- $ n $ — количество примеров.

Плюсы:
- Простота вычисления и интерпретации.
- Гладкая и дифференцируемая функция, что важно для градиентного спуска.
- Наказывает за большие ошибки сильнее, чем за маленькие.

Минусы:
- Чувствительность к выбросам (так как квадратичное увеличение ошибки делает влияние выбросов значительным).
- Предпочитает малые ошибки даже при наличии нескольких крупных ошибок.

Рекомендации:
- Используйте MSE, если выбросы не являются большой проблемой.
- Если в данных есть выбросы, рассмотрите **MAE** или **Huber Loss**.
- Хорошо работает в задачах, где важны небольшие отклонения предсказаний.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mse_loss = nn.MSELoss()
loss_value = mse_loss(y_pred, y_true)

f'MSE Loss: {loss_value.item()}'
```

# Root Mean Squared Error (RMSE)

RMSE (Root Mean Squared Error) — это корень из среднеквадратичной ошибки, популярная функция потерь для задач регрессии.

$$
RMSE = \sqrt{MSE}
$$

Плюсы:
- RMSE интерпретируется в тех же единицах, что и сами данные, что упрощает анализ.
- Подчеркивает большие ошибки за счет квадратичного роста.

Минусы:
- Чувствителен к выбросам, так как большие ошибки оказывают значительное влияние.
- Сложнее интерпретировать, чем MSE, из-за взятия корня.

Рекомендации:
- Используйте RMSE, если важны крупные ошибки, так как он штрафует их сильнее.
- Хорош для задач, где требуются интерпретируемые единицы измерения ошибки.
- Если выбросы являются проблемой, рассмотрите **MAE** или **Huber Loss**.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mse_loss = nn.MSELoss()
mse_value = mse_loss(y_pred, y_true)

rmse_value = torch.sqrt(mse_value)

f'RMSE Loss: {rmse_value.item()}'
```

# Mean Absolute Error (MAE)

MAE (Mean Absolute Error) — это средняя абсолютная ошибка, используемая в задачах регрессии.

$$
MAE = \frac{1}{n} \sum |y_i - \hat{y}_i|
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения,
- $ n $ — количество примеров.

Плюсы:
- Робастность к выбросам, так как ошибки не возводятся в квадрат.
- Легко интерпретируемый результат (средняя абсолютная разница между предсказанием и истиной).

Минусы:
- Недифференцируемая в точке 0, что может затруднять оптимизацию.
- Не акцентирует внимание на крупные ошибки (в отличие от MSE).

Рекомендации:
- Используйте MAE, если данные содержат выбросы.
- Хороший выбор, если важно минимизировать среднюю абсолютную разницу.
- Если нужно учитывать и небольшие, и крупные ошибки, рассмотрите **Huber Loss**.


```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mae_loss = nn.L1Loss()
loss_value = mae_loss(y_pred, y_true)

f'MAE Loss: {loss_value.item()}'
```

# Huber Loss

Huber Loss — это функция потерь, сочетающая в себе свойства **MSE** и **MAE**, что делает её более устойчивой к выбросам.

$$
L_{\delta}(a) =
\begin{cases}
\frac{1}{2} a^2, & \text{если } |a| \leq \delta \\
\delta (|a| - \frac{1}{2} \delta), & \text{если } |a| > \delta
\end{cases}
$$

Где:
- $ a = y_i - \hat{y}_i $ — разница между истинным и предсказанным значением,
- $ \delta $ — гиперпараметр, определяющий границу между **MSE** и **MAE**.

Плюсы:
- Менее чувствителен к выбросам по сравнению с **MSE**.
- Гладкая и дифференцируемая функция, что помогает при градиентном спуске.
- Объединяет преимущества **MSE** (малые ошибки) и **MAE** (устойчивость к выбросам).

Минусы:
- Требует подбора гиперпараметра $ \delta $.
- При неверном выборе $ \delta $ может работать хуже, чем **MSE** или **MAE**.

Рекомендации:
- Используйте **Huber Loss**, если данные содержат выбросы, но не в больших количествах.
- При очень больших выбросах рассмотрите **MAE** как альтернативу.
- Экспериментируйте с $ \delta $, но часто значения в диапазоне **1.0 – 2.0** работают хорошо.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

huber_loss = nn.HuberLoss(delta=1.0)
loss_value = huber_loss(y_pred, y_true)

f'Huber Loss: {loss_value.item()}'
```

# Log-Cosh Loss

Log-Cosh Loss — это функция потерь, которая ведет себя как **MSE** для маленьких ошибок, но как **MAE** для больших ошибок, что делает ее устойчивой к выбросам.

$$
L = \sum \log(\cosh(y_i - \hat{y}_i))
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения.

Плюсы:
- Менее чувствительна к выбросам, чем MSE.
- Гладкая и дифференцируемая функция.
- Ведет себя как MSE при малых ошибках.

Минусы:
- Вычислительно сложнее, чем MSE или MAE (из-за логарифма и гиперболического косинуса).
- В некоторых случаях может сходиться медленнее, чем MSE.

Рекомендации:
- Подходит для регрессионных задач, где есть выбросы.
- Если выбросы критичны, рассмотрите **Huber Loss**.
- Можно использовать как альтернативу MSE, если требуется более устойчивая функция потерь.


```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

def log_cosh_loss(y_pred, y_true):
    return torch.mean(torch.log(torch.cosh(y_pred - y_true)))

loss_value = log_cosh_loss(y_pred, y_true)

f'Log-Cosh Loss: {loss_value.item()}'
```
