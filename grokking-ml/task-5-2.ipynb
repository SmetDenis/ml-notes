Точка $ p $ классифицирована моделью неверно.

Модель использует функцию активации `step`, которая возвращает 1, если $ 2x_1 + 3x_2 \geq 4 $, и 0 в противном случае.

Проверим классификацию для точки $ p = (1, 1) $.

```python
def step_function(x):
    return 1 if x >= 0 else 0


# Координаты точки p
x1, x2 = 1, 1

# Вычисление линейной функции
output = 2 * x1 + 3 * x2 - 4

# Прогноз модели
prediction = step_function(output)

print("Классификация:", prediction)  # Ожидаем 0, так как 2*1 + 3*1 - 4 = 1
```

Ошибка перцептрона — это разница между фактическим значением и предсказанным.

```python
# Метка точки p
true_label = 0

# Ошибка перцептрона
error = true_label - prediction

print("Ошибка перцептрона:", error)
```

Корректируем веса, используя скорость обучения \(\eta = 0.01\).

```python
# Скорость обучения
eta = 0.01

# Изначальные веса
w1, w2, b = 2, 3, -4

# Обновление весов
w1 += eta * error * x1
w2 += eta * error * x2
b += eta * error

print("Новые веса:", w1, w2, b)
```

Прогноз новой моделью в точке $ p $.

```python
# Вычисление линейной функции с новыми весами
new_output = w1 * x1 + w2 * x2 + b

# Прогноз новой модели
new_prediction = step_function(new_output)

# Новая ошибка перцептрона
new_error = true_label - new_prediction

print("Новая классификация:", new_prediction)
print("Новая ошибка перцептрона:", new_error)
```
