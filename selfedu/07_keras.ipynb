```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow import keras
from tensorflow.keras.layers import Dense, Input
```

```python
x = np.array([-40, -10, 0, 8, 15, 22, 38])
y = np.array([-40, 14, 32, 46, 59, 72, 100])
```

```python
model = keras.Sequential([
    Input(shape=(1,)),  # Задание входных данных
    Dense(units=1, activation='linear')  # Линейный выход
])

model.compile(loss='mean_squared_error', optimizer=keras.optimizers.Adam(learning_rate=0.1))
```

```python
log = model.fit(x, y, epochs=1000, verbose=False)
```

```python
model.summary()
```

```python
model.predict(np.array([100, 0, 38]))
```

```python
model.get_weights()
```

```python
log.history['loss'][-1]
```

```python
plt.plot(log.history['loss'])
plt.grid(True)
plt.show()
```
