# Критерии качества нейронных сетей

## Распознавание

### 1. Hinge Loss (Функция хиндж)

**Описание**
Используется в задачах бинарной классификации, особенно в SVM. Наказывает ошибки сильнее, если предсказание далеко от истинного класса.

**Формула**
$$
L(y, \hat{y}) = \sum_{i} \max(0, 1 - y_i \hat{y}_i)
$$
где:
- $y_i \in \{-1, 1\}$ – истинная метка класса,
- $\hat{y}_i$ – предсказанный выход модели.

**Плюсы**
- Хорошо работает с разделимыми данными.
- Способствует увеличению разрыва между классами.

**Минусы**
- Не применяется для многоклассовой классификации.
- Неэффективен для плохо разделимых данных.

**Рекомендации**
Использовать в бинарной классификации, особенно в сочетании с SVM.

---

### 2. Binary Crossentropy (Бинарная кросс-энтропия)

**Описание**
Используется при бинарной классификации. Основана на теории вероятностей и измеряет расхождение между истинным распределением и предсказанным.

**Формула**
$$
L(y, \hat{y}) = -\frac{1}{N} \sum_{i} \left( y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right)
$$

**Плюсы**
- Интерпретируемые вероятности выхода.
- Хорошо работает с вероятностными моделями.

**Минусы**
- Чувствительна к несбалансированным данным.
- Требует использования сигмоидной функции активации.

**Рекомендации**
Использовать для бинарной классификации с вероятностными выходами.

---

### 3. Categorical Crossentropy (Категориальная кросс-энтропия)

**Описание**
Используется в задачах многоклассовой классификации.

**Формула**
$$
L(y, \hat{y}) = -\sum_{i} y_i \log(\hat{y}_i)
$$

где $y_i$ – вектор one-hot представления истинных классов.

**Плюсы**
- Подходит для многоклассовой классификации.
- Хорошо работает с вероятностными моделями.

**Минусы**
- Чувствительна к несбалансированным классам.
- Требует softmax на выходе.

**Рекомендации**
Использовать в задачах многоклассовой классификации.

## Обработка текста

### 4. Log-Cosh (Логарифмический гиперболический косинус)

**Описание**
Является гладкой версией MAE (Mean Absolute Error). Подходит для обработки текстов.

**Формула**
$$
L(y, \hat{y}) = \sum_{i} \log(\cosh(\hat{y}_i - y_i))
$$

**Плюсы**
- Обращается с выбросами мягче, чем MSE.
- Хорошо подходит для текстовых задач.

**Минусы**
- Медленнее, чем MSE.

**Рекомендации**
Использовать в задачах текстового анализа, когда выбросы возможны.

## Задачи регрессии

### 5. Mean Squared Error (Средний квадрат ошибок, MSE)

**Описание**
Классический метод оценки ошибки в регрессии.

**Формула**
$$
L(y, \hat{y}) = \frac{1}{N} \sum_{i} (y_i - \hat{y}_i)^2
$$

**Плюсы**
- Простота вычислений.
- Хорошо учитывает большие ошибки.

**Минусы**
- Чувствителен к выбросам.

**Рекомендации**
Использовать в регрессии, когда выбросов мало.

---

### 6. Mean Absolute Error (Средний модуль ошибок, MAE)

**Описание**
Среднее абсолютное отклонение предсказанных значений от истинных.

**Формула**
$$
L(y, \hat{y}) = \frac{1}{N} \sum_{i} |y_i - \hat{y}_i|
$$

**Плюсы**
- Менее чувствителен к выбросам, чем MSE.
- Интерпретируемый результат.

**Минусы**
- Не дифференцируем в точке 0.

**Рекомендации**
Использовать, если важны реальные ошибки в интерпретируемой форме.

---

### 7. Mean Absolute Percentage Error (Средний абсолютный процент ошибок, MAPE)

**Описание**
Оценивает ошибку в процентах, полезен в прогнозировании.

**Формула**
$$
L(y, \hat{y}) = \frac{1}{N} \sum_{i} \left| \frac{y_i - \hat{y}_i}{y_i} \right|
$$

**Плюсы**
- Подходит для прогноза временных рядов.
- Показывает относительную ошибку.

**Минусы**
- Не применим, если $y_i = 0$.
- Не подходит для очень маленьких значений.

**Рекомендации**
Использовать в прогнозировании, если важен процент ошибки.

---

### 8. Mean Squared Logarithmic Error (Средний квадрат логарифмических ошибок, MSLE)

**Описание**
Логарифмическая версия MSE, уменьшает влияние выбросов.

**Формула**
$$
L(y, \hat{y}) = \frac{1}{N} \sum_{i} \left( \log(1 + y_i) - \log(1 + \hat{y}_i) \right)^2
$$

**Плюсы**
- Стабилизирует ошибку при наличии выбросов.
- Учитывает относительные различия.

**Минусы**
- Не подходит для данных с нулевыми значениями.

**Рекомендации**
Использовать, если важна относительная ошибка при регрессии.
