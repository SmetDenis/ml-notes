#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
today = np.datetime64('today', 'D')  # Сегодня
yesterday = today - np.timedelta64(1, 'D')  # Вчера
tomorrow = today + np.timedelta64(1, 'D')  # Завтра

print("Yesterday:", yesterday)
print("Today:", today)
print("Tomorrow:", tomorrow)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A = np.ones(3) * 1
B = np.ones(3) * 2
np.add(A, B, out=B)
np.divide(A, 2, out=A)
np.negative(A, out=A)
np.multiply(A, B, out=A)
````
