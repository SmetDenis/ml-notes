## Оптимизация на основе моментов

**Описание**
Метод основан на накоплении экспоненциально сглаженной суммы предыдущих градиентов, что позволяет ускорить сходимость и уменьшить колебания при движении по "горным хребтам" функции потерь.

**Формулы**
$$
v_t = \gamma v_{t-1} + \eta \nabla_{\theta} J(\theta)
$$
$$
\theta = \theta - v_t
$$

**Плюсы**
- Ускоряет сходимость, особенно в глубоких долинах.
- Смягчает колебания в направлениях с сильными изменениями градиентов.

**Минусы**
- Параметр $\gamma$ (коэффициент моментума) требует тщательной настройки.
- Может сохранять "лишнюю инерцию", если значение $\gamma$ слишком велико.

**Рекомендации**
Применять, когда требуется стабилизировать процесс обучения и ускорить сходимость, особенно в задачах с зацепами и локальными минимумами.


## Ускоренные градиенты Нестерова

**Описание**
Метод Нестерова (NAG) улучшает подход momentum за счёт вычисления градиента в "предсказанном" положении, что позволяет точнее корректировать шаг обновления параметров.

**Формулы**
$$
v_t = \gamma v_{t-1} + \eta \nabla_{\theta} J\left(\theta - \gamma v_{t-1}\right)
$$
$$
\theta = \theta - v_t
$$

**Плюсы**
- Предсказывает будущее положение, что даёт более точное направление обновления.
- Часто приводит к более быстрой сходимости.

**Минусы**
- Вычислительно чуть сложнее обычного momentum.
- Также требует настройки параметра $\gamma$.

**Рекомендации**
Применять, когда важна точность направления обновления и требуется более агрессивное ускорение сходимости.

## Метод Adagrad

**Описание**
Адаптивный метод, который изменяет скорость обучения для каждого параметра на основе накопленной суммы квадратов градиентов. Особенно полезен для разреженных данных.

**Формулы**
$$
r_t = r_{t-1} + \nabla_{\theta} J(\theta)^2
$$
$$
\theta = \theta - \frac{\eta}{\sqrt{r_t + \epsilon}} \nabla_{\theta} J(\theta)
$$

**Плюсы**
- Автоматическая адаптация скорости обучения для каждого параметра.
- Хорошо работает с разреженными градиентами (например, в задачах обработки текста).

**Минусы**
- Скорость обучения может слишком быстро снижаться, что приведёт к преждевременной остановке обучения.

**Рекомендации**
Подходит для задач с разреженными данными, когда важна индивидуальная адаптация параметров.


## Методы RMSProp и Adadelta

**Описание**
Эти методы являются модификациями Adagrad, решающими проблему слишком быстрого уменьшения скорости обучения. Они используют экспоненциальное скользящее среднее квадратов градиентов.

**Формулы (RMSProp)**
$$
E[\nabla_{\theta} J(\theta)^2]_t = \gamma E[\nabla_{\theta} J(\theta)^2]_{t-1} + (1-\gamma)(\nabla_{\theta} J(\theta))^2
$$
$$
\theta = \theta - \frac{\eta}{\sqrt{E[\nabla_{\theta} J(\theta)^2]_t + \epsilon}} \nabla_{\theta} J(\theta)
$$

**Формулы (Adadelta)**
Adadelta устраняет необходимость фиксированного параметра скорости обучения:
$$
E[\nabla_{\theta} J(\theta)^2]_t = \gamma E[\nabla_{\theta} J(\theta)^2]_{t-1} + (1-\gamma)(\nabla_{\theta} J(\theta))^2
$$
$$
\Delta \theta_t = - \frac{\sqrt{E[\Delta \theta^2]_{t-1} + \epsilon}}{\sqrt{E[\nabla_{\theta} J(\theta)^2]_t + \epsilon}} \nabla_{\theta} J(\theta)
$$
$$
E[\Delta \theta^2]_t = \gamma E[\Delta \theta^2]_{t-1} + (1-\gamma)(\Delta \theta_t)^2
$$

**Плюсы**
- RMSProp стабилизирует скорость обучения, особенно при нестационарных задачах.
- Adadelta освобождает от необходимости вручную задавать скорость обучения.

**Минусы**
- Требуют дополнительной настройки гиперпараметра $\gamma$.
- Могут быть чувствительны к выбору $\epsilon$.

**Рекомендации**
Рекомендуются для задач с негладкими функциями потерь и для нейросетей, где требуется стабильная адаптация шага.


## Метод Adam и NAdam

**Описание**
Adam объединяет идеи momentum и RMSProp, используя адаптивные оценки первого (среднее) и второго (дисперсия) моментов градиента. NAdam — это модификация Adam с использованием поправок Нестерова.

**Формулы (Adam)**
$$
m_t = \beta_1 m_{t-1} + (1-\beta_1) \nabla_{\theta} J(\theta)
$$
$$
v_t = \beta_2 v_{t-1} + (1-\beta_2) \left(\nabla_{\theta} J(\theta)\right)^2
$$
$$
\hat{m}_t = \frac{m_t}{1-\beta_1^t}
$$
$$
\hat{v}_t = \frac{v_t}{1-\beta_2^t}
$$
$$
\theta = \theta - \frac{\eta}{\sqrt{\hat{v}_t} + \epsilon} \hat{m}_t
$$

**Формулы (NAdam)**
$$
m_t = \beta_1 m_{t-1} + (1-\beta_1) \nabla_{\theta} J(\theta)
$$
$$
\hat{m}_t = \frac{m_t}{1-\beta_1^t}
$$
$$
\theta = \theta - \frac{\eta}{\sqrt{\hat{v}_t} + \epsilon} \left(\beta_1 \hat{m}_t + \frac{(1-\beta_1)}{1-\beta_1^t} \nabla_{\theta} J(\theta)\right)
$$

**Плюсы**
- Высокая скорость сходимости.
- Адаптивность за счёт использования моментов первого и второго порядков.
- Хорошо работает в задачах с большим количеством параметров.

**Минусы**
- Требуют тщательной настройки гиперпараметров ($\beta_1$, $\beta_2$, $\eta$).
- В некоторых случаях могут приводить к переобучению, если параметры настроены неверно.

**Рекомендации**
Adam является популярным выбором для нейросетей и используется по умолчанию во многих библиотеках. NAdam рекомендуется, если требуется дополнительная коррекция за счёт Нестерова для ещё более точного направления обновления.
