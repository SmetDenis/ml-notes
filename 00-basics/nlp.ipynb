```python
import spacy

nlp = spacy.load('en_core_web_sm')

text = '''
🚀 Breaking: OpenAI just released GPT-5! 🤯🔥
💡 "The future of AI is now," says @sama.
📈 Stocks soar: $TSLA +5%, $AAPL -2% 📉
#AI #MachineLearning #TechNews
👉 Read more: https://openai.com/gpt-5
🗳️ Poll: Do you think AI will replace programmers? (Vote below 👇)
'''

print([str(token) for token in nlp(text.lower())])
```

```python
from nltk.tokenize import TweetTokenizer

tokenizer = TweetTokenizer()
draft_tokens = tokenizer.tokenize(text)
```

```python
def n_grams(text, n):
    return [text[i:i + n] for i in range(len(text) - n + 1)]

n_grams(draft_tokens, 3)
```

```python
text = u'The quick brown fox 🦊, named Mr. Jumps, jumped over 13 lazy dogs at 3:45 PM!'
for token in nlp(text):
    print('{} --> {} --> {}'.format(token, token.lemma_, token.pos_))
```

```python
for chunk in nlp(text).noun_chunks:
    print(chunk, '|', chunk.label_)
```
