```python
import torch
import torch.nn as nn
```

# Mean Squared Error (MSE)

MSE (Mean Squared Error) — это среднеквадратичная ошибка, широко используемая функция потерь для задач регрессии.

$$
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения,
- $ n $ — количество примеров.

Плюсы:
- Простота вычисления и интерпретации.
- Гладкая и дифференцируемая функция, что важно для градиентного спуска.
- Наказывает за большие ошибки сильнее, чем за маленькие.

Минусы:
- Чувствительность к выбросам (так как квадратичное увеличение ошибки делает влияние выбросов значительным).
- Предпочитает малые ошибки даже при наличии нескольких крупных ошибок.

Рекомендации:
- Используйте MSE, если выбросы не являются большой проблемой.
- Если в данных есть выбросы, рассмотрите **MAE** или **Huber Loss**.
- Хорошо работает в задачах, где важны небольшие отклонения предсказаний.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mse_loss = nn.MSELoss()
loss_value = mse_loss(y_pred, y_true)

f'MSE Loss: {loss_value.item()}'
```

# Root Mean Squared Error (RMSE)

RMSE (Root Mean Squared Error) — это корень из среднеквадратичной ошибки, популярная функция потерь для задач регрессии.

$$
RMSE = \sqrt{MSE}
$$

Плюсы:
- RMSE интерпретируется в тех же единицах, что и сами данные, что упрощает анализ.
- Подчеркивает большие ошибки за счет квадратичного роста.

Минусы:
- Чувствителен к выбросам, так как большие ошибки оказывают значительное влияние.
- Сложнее интерпретировать, чем MSE, из-за взятия корня.

Рекомендации:
- Используйте RMSE, если важны крупные ошибки, так как он штрафует их сильнее.
- Хорош для задач, где требуются интерпретируемые единицы измерения ошибки.
- Если выбросы являются проблемой, рассмотрите **MAE** или **Huber Loss**.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mse_loss = nn.MSELoss()
mse_value = mse_loss(y_pred, y_true)

rmse_value = torch.sqrt(mse_value)

f'RMSE Loss: {rmse_value.item()}'
```

# Mean Absolute Error (MAE)

MAE (Mean Absolute Error) — это средняя абсолютная ошибка, используемая в задачах регрессии.

$$
MAE = \frac{1}{n} \sum |y_i - \hat{y}_i|
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения,
- $ n $ — количество примеров.

Плюсы:
- Робастность к выбросам, так как ошибки не возводятся в квадрат.
- Легко интерпретируемый результат (средняя абсолютная разница между предсказанием и истиной).

Минусы:
- Недифференцируемая в точке 0, что может затруднять оптимизацию.
- Не акцентирует внимание на крупные ошибки (в отличие от MSE).

Рекомендации:
- Используйте MAE, если данные содержат выбросы.
- Хороший выбор, если важно минимизировать среднюю абсолютную разницу.
- Если нужно учитывать и небольшие, и крупные ошибки, рассмотрите **Huber Loss**.


```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

mae_loss = nn.L1Loss()
loss_value = mae_loss(y_pred, y_true)

f'MAE Loss: {loss_value.item()}'
```

# Huber Loss

Huber Loss — это функция потерь, сочетающая в себе свойства **MSE** и **MAE**, что делает её более устойчивой к выбросам.

$$
L_{\delta}(a) =
\begin{cases}
\frac{1}{2} a^2, & \text{если } |a| \leq \delta \\
\delta (|a| - \frac{1}{2} \delta), & \text{если } |a| > \delta
\end{cases}
$$

Где:
- $ a = y_i - \hat{y}_i $ — разница между истинным и предсказанным значением,
- $ \delta $ — гиперпараметр, определяющий границу между **MSE** и **MAE**.

Плюсы:
- Менее чувствителен к выбросам по сравнению с **MSE**.
- Гладкая и дифференцируемая функция, что помогает при градиентном спуске.
- Объединяет преимущества **MSE** (малые ошибки) и **MAE** (устойчивость к выбросам).

Минусы:
- Требует подбора гиперпараметра $ \delta $.
- При неверном выборе $ \delta $ может работать хуже, чем **MSE** или **MAE**.

Рекомендации:
- Используйте **Huber Loss**, если данные содержат выбросы, но не в больших количествах.
- При очень больших выбросах рассмотрите **MAE** как альтернативу.
- Экспериментируйте с $ \delta $, но часто значения в диапазоне **1.0 – 2.0** работают хорошо.

```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

huber_loss = nn.HuberLoss(delta=1.0)
loss_value = huber_loss(y_pred, y_true)

f'Huber Loss: {loss_value.item()}'
```

# Log-Cosh Loss

Log-Cosh Loss — это функция потерь, которая ведет себя как **MSE** для маленьких ошибок, но как **MAE** для больших ошибок, что делает ее устойчивой к выбросам.

$$
L = \sum \log(\cosh(y_i - \hat{y}_i))
$$

Где:
- $ y_i $ — истинные значения,
- $ \hat{y}_i $ — предсказанные значения.

Плюсы:
- Менее чувствительна к выбросам, чем MSE.
- Гладкая и дифференцируемая функция.
- Ведет себя как MSE при малых ошибках.

Минусы:
- Вычислительно сложнее, чем MSE или MAE (из-за логарифма и гиперболического косинуса).
- В некоторых случаях может сходиться медленнее, чем MSE.

Рекомендации:
- Подходит для регрессионных задач, где есть выбросы.
- Если выбросы критичны, рассмотрите **Huber Loss**.
- Можно использовать как альтернативу MSE, если требуется более устойчивая функция потерь.


```python
y_true = torch.tensor([3.0, -0.5, 2.0, 7.0])
y_pred = torch.tensor([2.5, 0.0, 2.0, 8.0])

def log_cosh_loss(y_pred, y_true):
    return torch.mean(torch.log(torch.cosh(y_pred - y_true)))

loss_value = log_cosh_loss(y_pred, y_true)

f'Log-Cosh Loss: {loss_value.item()}'
```

# Binary Cross-Entropy (Log Loss)

Binary Cross-Entropy (Log Loss) — это функция потерь, используемая в задачах бинарной классификации. Она измеряет разницу между предсказанными вероятностями и истинными метками классов.

$$
L = -\frac{1}{n} \sum \left[ y_i \log \hat{y}_i + (1 - y_i) \log (1 - \hat{y}_i) \right]
$$

Где:
- $ y_i $ — истинные метки (0 или 1),
- $ \hat{y}_i $ — предсказанные вероятности (от 0 до 1),
- $ n $ — количество примеров.

Плюсы:
- Подходит для задач бинарной классификации, где выходная вероятность интерпретируется как уверенность модели.
- Гладкая и дифференцируемая, что важно для оптимизации.
- Хорошо работает при сбалансированных данных.

Минусы:
- Чувствительна к несбалансированным классам (при сильном дисбалансе модель может просто предсказывать наиболее частый класс).
- Логарифм может приводить к числовым проблемам при предсказаниях, близких к 0 или 1.

Рекомендации:
- Используйте вместе с **sigmoid** на выходе нейросети.
- Для несбалансированных данных рассмотрите **Weighted BCE Loss** или **Focal Loss**.
- Хорошо работает в задачах, где важна уверенность в предсказаниях.

```python
y_true = torch.tensor([1, 0, 1, 1], dtype=torch.float32) # Истинные метки (0 или 1)
y_pred = torch.tensor([0.9, 0.1, 0.8, 0.7], dtype=torch.float32) # Предсказанные вероятности (от 0 до 1)

bce_loss = nn.BCELoss()
loss_value = bce_loss(y_pred, y_true)

f'BCE Loss: {loss_value.item()}'
```

# Categorical Cross-Entropy (CCE)

**Categorical Cross-Entropy** (CCE) — это функция потерь, используемая для многоклассовой классификации, где каждый образец принадлежит только одному классу.

$$
L = -\sum_{i=1}^{n} \sum_{j=1}^{k} y_{i,j} \log \hat{y}_{i,j}
$$

Где:
- $ y_{i,j} $ — истинная метка класса (one-hot encoding).
- $ \hat{y}_{i,j} $ — предсказанная вероятность для класса $ j $.
- $ k $ — количество классов.
- $ n $ — количество примеров.

Плюсы:
- Хорошо подходит для задач многоклассовой классификации.
- Основана на вероятностной интерпретации (работает с выходами softmax).
- Наказывает за низкую вероятность истинного класса, заставляя модель уверенно делать предсказания.

Минусы:
- Чувствительна к плохо откалиброванным вероятностям (если модель переобучена, может выдавать слишком уверенные предсказания).
- Не работает напрямую с индексами классов (используется one-hot encoding, но можно использовать `torch.nn.CrossEntropyLoss`, который принимает индексы).

Рекомендации:
- Используйте CCE в задачах классификации с более чем двумя классами.
- Всегда применяйте softmax перед вычислением вероятностей.
- Если классы несбалансированы, рассмотрите **Weighted Cross-Entropy** или **Focal Loss**.

```python
# Истинные метки (one-hot encoding)
y_true = torch.tensor([[0, 1, 0], [1, 0, 0], [0, 0, 1]])

# Предсказанные вероятности (выход softmax)
y_pred = torch.tensor([[0.1, 0.8, 0.1],
                       [0.7, 0.2, 0.1],
                       [0.1, 0.3, 0.6]])

loss_fn = nn.CrossEntropyLoss()

# CrossEntropyLoss в PyTorch ожидает индексы классов, а не one-hot encoding
y_true_indices = torch.tensor([1, 0, 2])

loss_value = loss_fn(y_pred, y_true_indices)

f'Categorical Cross-Entropy Loss: {loss_value.item()}'
```

# Sparse Categorical Cross-Entropy (Sparse CCE)

**Sparse Categorical Cross-Entropy (Sparse CCE)** — это функция потерь, используемая в многоклассовой классификации, где истинные метки представлены в виде индексов классов, а предсказания — как распределение вероятностей.

$$
L = -\frac{1}{n} \sum_{i=1}^{n} \log p_{i, y_i}
$$

Где:
- $ p_{i, y_i} $ — предсказанная вероятность истинного класса $ y_i $ для примера $ i $,
- $ n $ — количество примеров.

Плюсы:
- **Эффективность**: требует меньше памяти, так как истинные метки хранятся как индексы, а не как one-hot вектора.
- **Простота использования**: удобно при работе с данными, где классы представлены целыми числами.

Минусы:
- **Ограничение на входные данные**: истинные метки должны быть представлены как индексы, а не one-hot векторы.
- **Не подходит для многолейбловой классификации**: если у объекта несколько меток, лучше использовать Binary Cross-Entropy.

Рекомендации:
- Используйте Sparse CCE, если у вас большой набор классов и истинные метки представлены как индексы.
- Если данные уже в one-hot формате, используйте **Categorical Cross-Entropy**.
- Подходит для нейросетей с softmax-активацией на последнем слое.

```python
# Истинные метки (индексы классов)
y_true = torch.tensor([2, 0, 1])  # Классы: 2, 0, 1

# Предсказанные вероятности классов (логиты, выход из нейросети)
y_pred = torch.tensor([
    [1.0, 2.0, 3.0],  # Пример 1
    [3.0, 2.0, 1.0],  # Пример 2
    [0.5, 2.5, 1.5]   # Пример 3
])

loss_fn = nn.CrossEntropyLoss()
# Вычисляем Sparse CCE (автоматически применяется softmax к y_pred)
loss_value = loss_fn(y_pred, y_true)

f'Sparse Categorical Cross-Entropy Loss: {loss_value.item()}'
```

# Hinge Loss

Hinge Loss используется в задачах бинарной классификации, особенно в **Support Vector Machines (SVM)**. Она наказывает предсказания, которые находятся на неправильной стороне разделяющей гиперплоскости или слишком близки к ней.

$$
L = \frac{1}{n} \sum \max(0, 1 - y_i \hat{y}_i)
$$

Где:
- $ y_i \in \{-1, 1\} $ — истинные метки классов,
- $ \hat{y}_i $ — предсказанные значения,
- $ n $ — количество примеров.

Плюсы:
- Способствует максимальному разделению классов, что улучшает обобщающую способность модели.
- Игнорирует правильно классифицированные точки, если они находятся достаточно далеко от границы.
- Устойчивее к выбросам по сравнению с **Log Loss** (Cross-Entropy).

Минусы:
- Не дифференцируема в точке $ 1 - y_i \hat{y}_i = 0 $, что требует специальных оптимизационных методов.
- Подходит только для **линейно разделимых** данных; для нелинейных данных требуется использование ядровых методов.

Рекомендации:
- Используйте **Hinge Loss** в **SVM**, если данные хорошо разделимы гиперплоскостью.
- Если нужны вероятности, лучше применять **Cross-Entropy Loss**.
- В глубоких нейросетях часто заменяется **Squared Hinge Loss** для лучшей сходимости.

```python
# Истинные метки (должны быть -1 или 1)
y_true = torch.tensor([1, -1, 1, -1], dtype=torch.float32)

# Предсказанные значения (до порога, не вероятности)
y_pred = torch.tensor([0.8, -0.6, 1.2, -0.4], dtype=torch.float32)

# Определяем Hinge Loss
hinge_loss = nn.HingeEmbeddingLoss(margin=1.0)

# Вычисляем Hinge Loss
loss_value = hinge_loss(y_pred, y_true)

f'Hinge Loss: {loss_value.item()}'
```

# Squared Hinge Loss

Squared Hinge Loss — это модификация **Hinge Loss**, используемая в задачах классификации, особенно в SVM (Support Vector Machine). Она вычисляется по формуле:

$$
L = \frac{1}{n} \sum \max(0, 1 - y_i \hat{y}_i)^2
$$

Где:
- $ y_i \in \{-1, 1\} $ — истинные метки классов,
- $ \hat{y}_i $ — предсказанные значения (логиты),
- $ n $ — количество примеров.

Плюсы:
- Усиливает штраф за небольшие ошибки по сравнению с обычным Hinge Loss.
- Более гладкая функция, что делает градиентный спуск более стабильным.
- Способствует лучшему разделению классов за счет сильного наказания за малые отступы.

Минусы:
- Чувствительна к выбросам, так как квадратичное увеличение ошибки делает влияние выбросов значительным.
- Требует нормализации логитов, поскольку ожидает значения от $-1$ до $1$.

Рекомендации:
- Используйте Squared Hinge Loss в задачах линейной классификации, где важна строгая граница разделения классов.
- Если данные содержат выбросы, рассмотрите **Huber Loss** или **Log-Cosh Loss**.
- Работает лучше в моделях, основанных на SVM, но требует корректной нормализации данных.

```python
# Истинные метки классов (-1 или 1) и предсказанные логиты
y_true = torch.tensor([1, -1, 1, -1], dtype=torch.float32)
y_pred = torch.tensor([0.8, -0.5, 1.2, -1.5], dtype=torch.float32)

# Определяем Squared Hinge Loss
def squared_hinge_loss(y_pred, y_true):
    return torch.mean(torch.clamp(1 - y_true * y_pred, min=0) ** 2)

# Вычисляем Squared Hinge Loss
loss_value = squared_hinge_loss(y_pred, y_true)

f'Squared Hinge Loss: {loss_value.item()}'
```

Дополнительные:
* KL-Divergence (Kullback-Leibler Divergence) – мера различия между двумя распределениями вероятностей.
* Triplet Loss – используется в задаче поиска схожести объектов.
* Contrastive Loss – похожа на Triplet Loss, но работает с парами объектов.
* Focal Loss – модифицированная cross-entropy, уделяет больше внимания сложным образцам.
